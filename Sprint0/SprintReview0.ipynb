```
Flower myRose = new Flower("Роза", 45.5f, "Красный", 15);
Tree myOak = new Tree("Дуб", 350.0f, 12, "Лиственные");
Console.WriteLine("--- Наш сад создан ---");
Console.WriteLine($"В саду растет: {myRose.Type} \"{myRose.Name}\" (Высота: {myRose.Height} см, Цвет: {myRose.Color})");
Console.WriteLine($"В саду растет: {myOak.Type} \"{myOak.Name}\" (Высота: {myOak.Height} см, Возраст: {myOak.Age} лет)\n");
Console.WriteLine("--- Демонстрация методов и Полиморфизма ---");
myRose.Bloom();      
myRose.Grow(10.5f);
myRose.Wither();
Console.WriteLine();
myOak.Bloom();
myOak.Grow(50.0f);
myOak.GrowOlder(3);
public class Plant
{
    public string Name { get; set; }
    public string Type { get; set; }

    private float height;
    public float Height
    {
        get { return height; }
        set 
        {
            if (value < 0) height = 0; 
            else height = value;
        }
    }
    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }
    public virtual void Bloom()
    {
        Console.WriteLine($"Растение {Name} начинает цвести.");
    }
    public void Grow(float growthAmount)
    {
        if (growthAmount > 0)
        {
            Height += growthAmount;
            Console.WriteLine($"Растение {Name} выросло на {growthAmount} см. Текущая высота: {Height} см.");
        }
    }
}
public class Flower : Plant
{
    public string Color { get; set; }
    public int PetalsCount { get; set; }
    public Flower(string name, float height, string color, int petalsCount) 
        : base(name, "Цветок", height)
    {
        Color = color;
        PetalsCount = petalsCount;
    }
    public override void Bloom()
    {
        Console.WriteLine($"Цветок {Name} распускается и радует глаз!");
    }
    public void Wither()
    {
        Console.WriteLine($"Цветок {Name} начинает увядать...");
    }
}
public class Tree : Plant
{
    public int Age { get; set; }
    public string LeavesType { get; set; }
    public Tree(string name, float height, int age, string leavesType) 
        : base(name, "Дерево", height)
    {
        Age = age;
        LeavesType = leavesType;
    }
    public override void Bloom()
    {
        Console.WriteLine($"Дерево {Name} покрывается новыми листьями!");
    }
    public void GrowOlder(int years)
    {
        if (years > 0)
        {
            Age += years;
            Console.WriteLine($"Дерево {Name} стало старше на {years} г. Новый возраст: {Age} лет."); 
        }
    }
}
```
