```
Computer myPC = ComputerFactory.CreateGamingPC();
myPC.ShowConfiguration();
RAM oldRam = new RAM("Kingston FURY 32GB", 15000m, 32);
GPU rtx4090 = new GPU("NVIDIA RTX 4090", 220000m, 24);
CPU i9 = new CPU("Intel Core i9-14900K", 65000m, 24);
Console.WriteLine($"[Операторы]: Цена RTX 4090 равна цене Core i9? -> {(rtx4090 == i9 ? "Да" : "Нет")}");
HDD extraHdd = new HDD("WD Blue 1TB", 4500m, 1);
myPC.AddComponent(extraHdd);
myPC.RemoveComponent(extraHdd);
myPC.ShowConfiguration();
public interface IInstallable
{
    void Install();
}
public abstract class Component
{
    public string Name { get; set; }
    private decimal _price;
    public decimal Price
    {
        get { return _price; }
        set { if (value < 0) _price = 0; else _price = value; }
    }
    protected Component(string name, decimal price)
    {
        Name = name;
        Price = price;
    }
    public virtual string GetInfo()
    {
        return $"Компонент: {Name} | Цена: {Price} руб.";
    }
    public override string ToString()
    {
        return GetInfo();
    }
    public static bool operator ==(Component c1, Component c2)
    {
        if (ReferenceEquals(c1, c2)) return true;
        if (c1 is null || c2 is null) return false;
        return c1.Price == c2.Price;
    }
    public static bool operator !=(Component c1, Component c2)
    {
        return !(c1 == c2);
    }
    public override bool Equals(object obj)
    {
        if (obj is Component other) return this == other;
        return false;
    }
    public override int GetHashCode()
    {
        return Price.GetHashCode();
    }
}
public class CPU : Component, IInstallable
{
    public int Cores { get; set; }
    public CPU(string name, decimal price, int cores) : base(name, price) 
    { 
        Cores = cores; 
    }
    public override string GetInfo() 
    {
        return base.GetInfo() + $" [CPU | Ядер: {Cores}]";
    }
    public void Install() 
    {
        Console.WriteLine($"Процессор {Name} аккуратно установлен в сокет материнской платы.");
    }
}
public class RAM : Component, IInstallable
{
    public int SizeGB { get; set; }
    public RAM(string name, decimal price, int sizeGB) : base(name, price) 
    { 
        SizeGB = sizeGB; 
    }
    public override string GetInfo() 
    {
        return base.GetInfo() + $" [RAM | Объем: {SizeGB} GB]";
    }
    public void Install() 
    {
        Console.WriteLine($"Модуль памяти {Name} защелкнут в DIMM-слот.");
    }
}
public class GPU : Component, IInstallable
{
    public int VramGB { get; set; }
    public GPU(string name, decimal price, int vramGB) : base(name, price) 
    { 
        VramGB = vramGB; 
    }
    public override string GetInfo() 
    {
        return base.GetInfo() + $" [GPU | Видеопамять: {VramGB} GB]";
    }
    public void Install() 
    {
        Console.WriteLine($"Видеокарта {Name} вставлена в слот PCIe и подключена к питанию.");
    }
}
public class HDD : Component, IInstallable
{
    public int CapacityTB { get; set; }
    public HDD(string name, decimal price, int capacityTB) : base(name, price) 
    { 
        CapacityTB = capacityTB; 
    }
    public override string GetInfo() 
    {
        return base.GetInfo() + $" [HDD | Емкость: {CapacityTB} TB]";
    }
    public void Install() 
    {
        Console.WriteLine($"Жесткий диск {Name} закреплен в корзине корпуса и подключен по SATA.");
    }
}

public class Computer
{
    private List<Component> _components = new List<Component>();
    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        _components.Add(component);
        Console.WriteLine($"[Добавление]: {component.Name} добавлен в конфигурацию ПК.");
        component.Install();
    }
    public void RemoveComponent<T>(T component) where T : Component
    {
        if (_components.Remove(component))
        {
            Console.WriteLine($"[Удаление]: {component.Name} извлечен из компьютера.");
        }
    }
    public decimal GetTotalPrice()
    {
        decimal total = 0;
        foreach (var comp in _components) total += comp.Price;
        return total;
    }
    public void ShowConfiguration()
    {
        Console.WriteLine("\n---Текущая конфигурация компьютера ---");
        foreach (var comp in _components)
        {
            Console.WriteLine($" • {comp}");
        }
        Console.WriteLine($"Общая стоимость сборки: {GetTotalPrice()} руб.\n");
    }
}
public static class ComputerFactory
{
    public static Computer CreateGamingPC()
    {
        Computer pc = new Computer();
        Console.WriteLine("[Фабрика]: Запущена сборка мощного Игрового ПК...");
        pc.AddComponent(new CPU("Intel Core i9-14900K", 65000m, 24));
        pc.AddComponent(new RAM("Kingston FURY 32GB", 15000m, 32));
        pc.AddComponent(new GPU("NVIDIA RTX 4090", 220000m, 24));
        pc.AddComponent(new HDD("Seagate BarraCuda 2TB", 7000m, 2)); 
        return pc;
    }
}
```
