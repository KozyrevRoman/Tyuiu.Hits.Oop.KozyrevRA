<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс CreditCard в C#, который будет представлять информацию 
о кредитных картах. На основе этого класса разработать 2-3 производных класса, 
демонстрирующих принципы наследования и полиморфизма. В каждом из классов 
должны быть реализованы новые атрибуты и методы, а также переопределены 
некоторые методы базового класса для демонстрации полиморфизма. 
Требования к базовому классу CreditCard: 
• Атрибуты: Номер карты (CardNumber), Холдера (HolderName), Срок 
действия (ExpiryDate). 
• Методы: 
o GetInfo(): метод для получения информации о кредитной карте в виде 
строки. 
o Pay(): метод для оплаты покупки с использованием карты. 
o CheckBalance(): метод для проверки баланса на карте. 
Требования к производным классам: 
1. GoldCard (GoldCreditCard): Должен содержать дополнительные атрибуты, 
такие как Бесплатные бонусные мили (BonusMiles). Метод Pay() должен быть 
переопределен для добавления информации о получении бонусных миль 
при оплате покупки. 
2. PremiumCard (PremiumCreditCard): Должен содержать дополнительные 
атрибуты, 
такие 
как 
Ассистент 
поддержки 
(SupportAssistant). 
Метод CheckBalance() должен быть переопределен для предоставления 
возможности обратиться за помощью к ассистенту поддержки в случае 
проблем с балансом. 
3. CorporateCard (CorporateCreditCard) (если требуется третий класс): Должен 
содержать дополнительные атрибуты, такие как Компания (Company). 
Метод GetInfo() должен быть переопределен для включения информации о 
компании в описании карты.

Дополнительное задание.

Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

```
GoldCreditCard gold = new GoldCreditCard("4444-5555-6666-7777", "Иван", "12/28", 50000, 100);
Console.WriteLine(gold.GetInfo());
gold.CheckBalance();
gold.Pay(3700);
Console.WriteLine("---------------------------------------------");
PremiumCreditCard premium = new PremiumCreditCard("1111-2222-3333-4444", "Петр", "05/30", 150000, "Олег");
Console.WriteLine(premium.GetInfo());
premium.CheckBalance();
premium.Pay(25000);
Console.WriteLine("---------------------------------------------");
CorporateCreditCard corporate = new CorporateCreditCard("7777-8888-9999-0000", "Анна", "09/27", 1000000, "ООО \"Tiger-Soft\"");
Console.WriteLine(corporate.GetInfo());
corporate.CheckBalance();
corporate.Pay(5000);
Console.WriteLine("---------------------------------------------");
BankTerminal terminal = new BankTerminal();
terminal.ProcessPayment(gold, 1000);
Console.WriteLine("---------------------------------------------");

gold.Pay(1500, "Супермаркеты");
CardHolder<GoldCreditCard> goldWallet = new CardHolder<GoldCreditCard>();
goldWallet.AddCard(gold);
goldWallet.ShowAllCards();
CardHolder<CreditCard> centralWallet = new CardHolder<CreditCard>();
centralWallet.AddCard(premium);
centralWallet.AddCard(corporate);
centralWallet.ShowAllCards();
public class CardHolder<T> where T : CreditCard
{
    private List<T> _myCards = new List<T>();

    public void AddCard(T card)
    {
        _myCards.Add(card);
        Console.WriteLine($"[Generic Кошелек]: Добавлена карта владельца {card.HolderName}");
    }

    public void ShowAllCards()
    {
        Console.WriteLine($"\n--- Список карт в кошельке типа {typeof(T).Name} ---");
        foreach (var card in _myCards)
        {
            Console.WriteLine(card.GetInfo());
        }
    }
}
public class CreditCard
{
    public string CardNumber { get; set; }
    public string HolderName { get; set; }
    public string ExpiryDate { get; set; }
    public string Status { get; set; } = "Активна";

    private double _balance;
    public double Balance
    {
        get { return _balance; }
        set
        {
            if (value < 0) _balance = 0;
            else _balance = value;
        }
    }
    public CreditCard(string cardNumber, string holderName, string expiryDate, double balance)
    {
        CardNumber = cardNumber;
        HolderName = holderName;
        ExpiryDate = expiryDate;
        Balance = balance;
    }
    public virtual string GetInfo()
    {
        return $"Карта: {CardNumber} | Владелец: {HolderName} | Срок: {ExpiryDate} | Статус: {Status}";
    }
    public virtual void Pay(double amount)
    {
        Balance -= amount;
        Console.WriteLine($"Списано {amount} руб. Остаток: {Balance} руб.");
    }
    public void Pay(double amount, string category)
    {
        Balance -= amount;
        Console.WriteLine($"Списано {amount} руб. в категории '{category}'. Остаток: {Balance} руб.");
    }

    public virtual void CheckBalance()
    {
        Console.WriteLine($"Текущий баланс карты: {Balance} руб.");
    }
}
public class GoldCreditCard : CreditCard
{
    public int BonusMiles { get; set; }
    public GoldCreditCard(string cardNumber, string holderName, string expiryDate, double balance, int miles)
        : base(cardNumber, holderName, expiryDate, balance)
    {
        BonusMiles = miles;
    }
    public override void Pay(double amount)
    {
        base.Pay(amount);
        int plusMiles = (int)(amount / 1000);
        BonusMiles += plusMiles;
        Console.WriteLine($"-> [Gold Бонус] Начислено {plusMiles} миль! Всего миль: {BonusMiles}");
    }
}
public class PremiumCreditCard : CreditCard
{
    public string SupportAssistant { get; set; }
    public PremiumCreditCard(string cardNumber, string holderName, string expiryDate, double balance, string assistant)
        : base(cardNumber, holderName, expiryDate, balance)
    {
        SupportAssistant = assistant;
    }
    public override void Pay(double amount)
    {
        base.Pay(amount);
        Console.WriteLine($"-> [Premium Статус] Transaction обработана вне очереди.");
    }
    public override void CheckBalance()
    {
        base.CheckBalance();
        Console.WriteLine($"-> [Premium Поддержка] Ваш личный ассистент ({SupportAssistant}) на связи.");
    }
}
public class CorporateCreditCard : CreditCard
{
    public string Company { get; set; }
    public CorporateCreditCard(string cardNumber, string holderName, string expiryDate, double balance, string company)
        : base(cardNumber, holderName, expiryDate, balance)
    {
        Company = company;
    }
    public override void Pay(double amount)
    {
        base.Pay(amount);
        Console.WriteLine($"-> [Corporate Отчет] Расход зафиксирован по счету компании ({Company}).");
    }
    public override string GetInfo()
    {
        return base.GetInfo() + $" | Компания: {Company}";
    }
}
public class BankTerminal
{
    public void ProcessPayment(CreditCard card, double amount)
    {
        Console.WriteLine($"\n[Терминал принимает карту {card.HolderName}]..."); 
        card.Pay(amount);
    }
}
```
