```
Store techStore = new Store("М.Видео", "ТЦ 'Кристалл', 2 этаж", "09:00 - 21:00");
Seller activeSeller = new Seller("Антон", "EMP-404", 55000.00m, "anton_pro@mvideo.ru");
Product console = new Product("PlayStation 5", 55000.00m, 5, "Консоли", "Игровая приставка Slim 1TB");
Product gamepad = new Product("Геймпад DualSense", 7990.00m, 12, "Аксессуары", "Беспроводной контроллер (Белый)");
techStore.AddSeller(activeSeller);
activeSeller.AddProduct(console);
activeSeller.AddProduct(gamepad);
Console.WriteLine("\n=======================================================");
Console.WriteLine($"   ДОБРО ПОЖАЛОВАТЬ В МАГАЗИН: {techStore.GetStoreInfo()}");
Console.WriteLine("=======================================================");
Console.WriteLine($"На смене сегодня: {activeSeller.GetSellerInfo()}");
Console.WriteLine("\n--- Проверка складского учета (Методы Product) ---");
Console.WriteLine(console.GetProductInfo());
Console.WriteLine($"Общая стоимость всех PS5 на складе: {console.GetTotalPrice()} руб.");
Console.WriteLine("\n[Система]: На склад привезли новую партию геймпадов (+5 шт).");
gamepad.UpdateQuantity(5); 
techStore.ListProducts();
Console.WriteLine("\n[Покупатель]: 'Привет! Я хочу купить одну приставку и два геймпада'.");
Console.WriteLine("\n--- Оформление покупки на кассе ---");
activeSeller.SellProduct(console, 1);  
activeSeller.SellProduct(gamepad, 2);  
Console.WriteLine("\n[Покупатель]: 'Ого, а можно мне со склада сразу еще 10 приставок забрать для друзей?'");
activeSeller.SellProduct(console, 10); 
techStore.ListProducts();public class Product
{
    private string name;
    private decimal price;
    private int quantity;
    protected string category;
    internal string description;
    public Product(string name, decimal price, int quantity, string category, string description)
    {
        this.name = name;
        this.price = price;
        Quantity = quantity;
        this.category = category;
        this.description = description;
    }
    public string Name { get { return name; } set { name = value; } }
    public decimal Price { get; set; }
    
    public int Quantity
    {
        get { return quantity; }
        set { if (value >= 0) quantity = value; else quantity = 0; }
    }
    public decimal GetTotalPrice()
    {
        return price * quantity;
    }
    public void UpdateQuantity(int amount)
    {
        Quantity += amount;
    }
    public string GetProductInfo()
    {
        return $"Товар: {name} | Категория: {category} | Цена: {price} руб. | Кол-во: {quantity} шт. | Описание: {description}";
    }
}
public class Seller
{
    private string name;
    private string employeeId;
    protected decimal salary;
    internal string contactInfo;
    private List<Product> sellerProducts = new List<Product>();
    public Seller(string name, string employeeId, decimal salary, string contactInfo)
    {
        this.name = name;
        this.employeeId = employeeId;
        this.salary = salary;
        this.contactInfo = contactInfo;
    }
    public void AddProduct(Product product)
    {
        sellerProducts.Add(product);
        Console.WriteLine($"Продавец {name} взял под контроль товар: {product.Name}");
    }
    public void SellProduct(Product product, int quantity)
    {
        if (product.Quantity >= quantity)
        {
            product.UpdateQuantity(-quantity);
            Console.WriteLine($"Продавец {name} продал {quantity} шт. товара {product.Name}. Осталось на складе: {product.Quantity} шт.");
        }
        else
        {
            Console.WriteLine($"Продавец {name} не может продать {quantity} шт. товара {product.Name}. На складе всего {product.Quantity} шт.");
        }
    }
    public string GetSellerInfo()
    {
        return $"Продавец: {name} | ID: {employeeId} | Контакты: {contactInfo}";
    }
    public List<Product> GetProducts()
    {
        return sellerProducts;
    }
}
public class Store
{
    private string storeName;
    private string location;
    public string storeHours;
    private List<Seller> storeSellers = new List<Seller>();
    public Store(string storeName, string location, string storeHours)
    {
        this.storeName = storeName;
        this.location = location;
        this.storeHours = storeHours;
    }
    public void AddSeller(Seller seller)
    {
        storeSellers.Add(seller);
        Console.WriteLine($"В магазин '{storeName}' устроен новый сотрудник.");
    }
    public void ListProducts()
    {
        Console.WriteLine($"\n--- Список всех товаров в магазине '{storeName}' ---");
        bool hasProducts = false; 
        foreach (var seller in storeSellers)
        {
            foreach (var product in seller.GetProducts())
            {
                Console.WriteLine(product.GetProductInfo());
                hasProducts = true;
            }
        }
        if (!hasProducts) Console.WriteLine("В магазине пока нет товаров.");
    }
    public string GetStoreInfo()
    { 
        return $"Магазин: \"{storeName}\" | Адрес: {location} | Часы работы: {storeHours}";
    }
}
```
