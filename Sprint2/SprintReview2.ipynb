```python
using System;
using System.Collections.Generic;

// =======================================================================
// 4. МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ: Интерфейсы
// =======================================================================
public interface IReportable
{
    string GenerateReport(); // Метод возвращает строку с отчетом
}

public interface ITeamLeader
{
    string[] GetTeamMembers(); // Метод возвращает массив строк с членами команды
}

// =======================================================================
// 1. Базовый класс Employee (Сотрудник)
// =======================================================================
public class Employee
{
    public string Name { get; set; }
    public string Position { get; set; }
    
    // Скрытое поле и умный сеттер для защиты зарплаты от отрицательных значений
    private float _salary;
    public float Salary
    {
        get { return _salary; }
        set
        {
            if (value < 0) _salary = 0; // Защита от дурака
            else _salary = value;
        }
    }

    // Конструктор базового класса
    public Employee(string name, string position, float salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }

    // Виртуальный метод вывода инфо (чтобы дочерние классы могли делать override)
    public virtual void PrintInfo()
    {
        Console.WriteLine($"Сотрудник: {Name} | Должность: {Position} | Зарплата: {Salary} руб.");
    }
}

// =======================================================================
// 2. ПРОСТОЕ НАСЛЕДОВАНИЕ: Класс Manager (Менеджер)
// =======================================================================
public class Manager : Employee
{
    public string Department { get; set; } // Отдел управления

    // Конструктор менеджера прокидывает общие данные наверх родителю через : base()
    public Manager(string name, string position, float salary, string department) 
        : base(name, position, salary)
    {
        Department = department;
    }

    // Переопределение метода вывода информации
    public override void PrintInfo()
    {
        base.PrintInfo(); // Вызываем базовый вывод родителя
        Console.WriteLine($"-> Отдел управления: {Department}");
    }
}

// =======================================================================
// 3 и 4. СЛОЖНОЕ И МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ: Класс ProjectManager
// (Иерархия: Employee -> Manager -> ProjectManager + 2 Интерфейса)
// =======================================================================
public class ProjectManager : Manager, IReportable, ITeamLeader
{
    public string ProjectName { get; set; } // Название проекта

    // Конструктор проектного менеджера
    public ProjectManager(string name, float salary, string department, string projectName) 
        : base(name, "Проектный менеджер", salary, department)
    {
        ProjectName = projectName;
    }

    // Переопределение метода PrintInfo()
    public override void PrintInfo()
    {
        base.PrintInfo(); // Вызывает цепочку: Manager -> Employee
        Console.WriteLine($"-> Закрепленный проект: {ProjectName}");
    }

    // Реализация метода интерфейса IReportable
    public string GenerateReport()
    {
        return $"[ОТЧЕТ ПО ПРОЕКТУ '{ProjectName}']: Все задачи спринта выполнены в срок. Багов не обнаружено.";
    }

    // Реализация метода интерфейса ITeamLeader
    public string[] GetTeamMembers()
    {
        // Возвращаем массив членов команды, как просили в задании
        return new string[] { "Разработчик Антон", "Дизайнер Анна", "Тестировщик Максим" };
    }
}

// =======================================================================
// 5. ТЕСТИРОВАНИЕ СИСТЕМЫ
// =======================================================================

Console.WriteLine("=== Проверка корпоративной структуры предприятия ===\n");

// 1. Создаем экземпляры каждого класса
Employee regularEmp = new Employee("Олег", "Младший разработчик", 60000f);
Manager deptManager = new Manager("Елена", "Начальник IT-отдела", 120000f, "Департамент разработки");
ProjectManager projManager = new ProjectManager("Антон", 150000f, "Департамент разработки", "Система 'Tiger-Soft'");

// 2. Вызываем метод PrintInfo() для каждого экземпляра
Console.WriteLine("--- Информация о сотрудниках ---");
regularEmp.PrintInfo();
Console.WriteLine("---------------------------------------------");
deptManager.PrintInfo();
Console.WriteLine("---------------------------------------------");
projManager.PrintInfo();
Console.WriteLine("---------------------------------------------");

// 3. Тестируем реализованные методы интерфейсов в ProjectManager
Console.WriteLine("\n--- Проверка работы интерфейсов ProjectManager ---");

// Вызываем метод генерации отчета (IReportable)
string finalReport = projManager.GenerateReport();
Console.WriteLine(finalReport);

// Вызываем метод получения списка команды (ITeamLeader)
Console.WriteLine($"\n👥 Члены команды под руководством {projManager.Name}:");
string[] team = projManager.GetTeamMembers();
foreach (string member in team)
{
    Console.WriteLine($" - {member}");
}
```
